```python
import matplotlib.pyplot as plt
import pandas as pd
```

# JURNAL

### SKENARIO 1

```python
guided_df = pd.read_csv("../hasil/record/GA-GUIDED.csv")
traditional_df = pd.read_csv("../hasil/record/GA-NOHELPER.csv")
gwo_ga_guided_df = pd.read_csv("../hasil/record/GA-GWO15-8.csv")
```

```python
gwo_ga_guided_df
```

```python
traditional_df
```

```python
guided_df
```

```python
plt.figure(figsize=(10,8))
plt.plot(
    traditional_df['iterasi'],
    traditional_df['best'],
    label='Traditional GA',
    linewidth=2
)

plt.plot(
    guided_df['iterasi'],
    guided_df['best'],
    label='Guided Operators GA',
    linewidth=2
)

plt.plot(
    gwo_ga_guided_df['iterasi'],
    gwo_ga_guided_df['best'],
    label='GA-GWO',
    linewidth=2
)

plt.xlabel('Iteration')
plt.ylabel('Best Fitness')
plt.title('Best Fitness Convergence Comparison')
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,8))
plt.plot(
    traditional_df['iterasi'],
    traditional_df['avg'],
    label='Traditional GA',
    linewidth=2
)

plt.plot(
    guided_df['iterasi'],
    guided_df['avg'],
    label='Guided Operators GA',
    linewidth=2
)

plt.plot(
    gwo_ga_guided_df['iterasi'],
    gwo_ga_guided_df['avg'],
    label='GA-GWO',
    linewidth=2
)

plt.xlabel('Iteration')
plt.ylabel('Average Fitness')
plt.title('Average Fitness Convergence Comparison')
plt.legend()
plt.grid(True)

plt.show()
```
